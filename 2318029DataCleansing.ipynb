# 📒Cleansing Data Olimpiade

# **1. Import Library & Load Dataset**
Kita mulai dengan mengimpor pustaka `pandas` dan membaca dataset Olimpiade.


```python
import pandas as pd
from google.colab import drive

#mount drive
drive.mount('/content/drive')

#path file
file_path = "/content/drive/MyDrive/Colab Notebooks/Summer-Olympic-medals-1976-to-2008.csv"
df = pd.read_csv(file_path, encoding="latin1")

# Tampilkan 5 baris pertama
df.head()
```

# **2. Cek Missing Values**
Kita periksa apakah ada data yang kosong (missing values) pada setiap kolom.


```python
df.isnull().sum()
```

# **3. Cek & Hapus Duplikasi**
Cek apakah ada baris data yang terduplikasi. Jika ada, kita hapus agar data bersih.


```python
print("Jumlah data duplikat:", df.duplicated().sum())
df = df.drop_duplicates()
```

# **4. Ubah Format Tanggal**
Kolom `Year` masih bertipe float, kita ubah menjadi integer agar lebih rapi.


```python
df['Year'] = df['Year'].astype('Int64')
df['Year'].head()
```

# **5. Normalisasi Data Teks**
Agar konsisten, kita ubah semua teks pada kolom bertipe object menjadi huruf kecil (lowercase)
dan hilangkan spasi di awal/akhir.


```python
for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].str.strip().str.lower()
df.head()
```

# ** 6. Tangani Nilai Negatif**
Kita cek semua kolom numerik, apakah ada nilai negatif. Jika ada, kita ganti dengan `NaN`.


```python
num_cols = df.select_dtypes(include=['int64','float64']).columns
for col in num_cols:
    negatif = df[df[col] < 0]
    if not negatif.empty:
        print(f"Ada nilai negatif di kolom {col}, diganti jadi NaN")
        df.loc[df[col] < 0, col] = None
```

# **7. Hapus Kolom Tidak Relevan**
Kolom `Country_Code` dan `Event_gender` tidak terlalu diperlukan, jadi kita hapus.


```python
df = df.drop(columns=['Country_Code', 'Event_gender'], errors='ignore')
df.head()
```

# **8. Dataset Setelah Cleansing**
Kita lihat struktur data (`info`) dan 5 baris pertama setelah proses cleansing.


```python
df.info()
df.head()
```

# **9. Data Enrichment**


---




```python
# Mapping Medal ke angka
medal_score_map = {"gold": 3, "silver": 2, "bronze": 1}
df["Medal_Score"] = df["Medal"].map(medal_score_map).astype("Int64")

df[["Medal", "Medal_Score"]].head()
```

```python
# 1. Medal_Score
medal_score_map = {"gold": 3, "silver": 2, "bronze": 1}
df["Medal_Score"] = df["Medal"].map(medal_score_map).astype("Int64")

# 2. Decade (dekade dari tahun)
df["Decade"] = ((df["Year"] // 10) * 10).astype("Int64")

# 3. Event_Type (individu vs tim, rule sederhana)
df["Event_Type"] = df["Event"].apply(
    lambda x: "team" if isinstance(x, str) and ("team" in x or "relay" in x) else "individual"
)


# 4. Buat ringkasan agregasi per negara
country_summary = df.groupby("Country").agg(
    Total_Medals=("Medal", "count"),
    Gold=("Medal", lambda x: (x == "gold").sum()),
    Silver=("Medal", lambda x: (x == "silver").sum()),
    Bronze=("Medal", lambda x: (x == "bronze").sum()),
    Total_Score=("Medal_Score", "sum")
).reset_index()


# Tampilkan hasil enrichment
df.head(), country_summary.head()
```
